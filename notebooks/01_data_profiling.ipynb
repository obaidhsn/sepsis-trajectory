# 01 · Data profiling
Reproduce the dataset numbers in the proposal (class balance, missingness, label timing).

```python
import sys; sys.path.insert(0, '../src')
import numpy as np, pandas as pd
from sepsis.data import load_raw_patients
from sepsis.constants import DYNAMIC, LABEL

raw = load_raw_patients('../data/raw')
len(raw)
```

```python
df = pd.concat([d.assign(patient=p, hospital=h) for p, (h, d) in raw.items()])
print('hours:', len(df))
print('septic patients:', df.groupby('patient')[LABEL].max().sum())
print('positive hours %:', 100 * df[LABEL].mean())
```

```python
missing = df[DYNAMIC].isna().mean().sort_values()
missing.plot.barh(figsize=(6, 9), title='Fraction of hours missing');
```

```python
# Label timing: hours from first positive label to end of record (leakage check)
first_pos = df[df[LABEL] == 1].groupby('patient').ICULOS.min()
last = df.groupby('patient').ICULOS.max()
(last[first_pos.index] - first_pos).describe()
```
